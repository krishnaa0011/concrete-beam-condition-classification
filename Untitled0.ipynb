```python
# ============================================================
# CONCRETE BEAM CONDITION CLASSIFICATION USING RANDOM FOREST
# CSV Creation + AI Training + Evaluation + Graphs
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    ConfusionMatrixDisplay
)

# 1. CREATE DEMO CSV DATASET
# IMPORTANT: These are synthetic/demo values, not real sensor data.

np.random.seed(42)

number_per_condition = 250
rows = []
time_ms = 0

for condition in ["Healthy", "Abnormal"]:

    for i in range(number_per_condition):

        if condition == "Healthy":
            accel_x = np.random.normal(0.05, 0.12)
            accel_y = np.random.normal(-0.03, 0.12)
            accel_z = np.random.normal(9.78, 0.18)
            vibration = int(np.random.random() < 0.08)

        else:
            accel_x = np.random.normal(0.20, 0.42)
            accel_y = np.random.normal(-0.12, 0.40)
            accel_z = np.random.normal(9.65, 0.50)
            vibration = int(np.random.random() < 0.55)

        # Acceleration magnitude (m/s^2)
        magnitude = np.sqrt(
            accel_x**2 + accel_y**2 + accel_z**2
        )

        rows.append({
            "time_ms": time_ms,
            "accel_x": round(accel_x, 4),
            "accel_y": round(accel_y, 4),
            "accel_z": round(accel_z, 4),
            "accel_magnitude": round(magnitude, 4),
            "vibration_do": vibration,
            "condition": condition
        })

        time_ms += 100

# Convert to DataFrame and shuffle rows
df = pd.DataFrame(rows)
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

# Save CSV
csv_filename = "concrete_beam_dataset.csv"
df.to_csv(csv_filename, index=False)

print("CSV DATASET CREATED SUCCESSFULLY")
print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

print("\nCondition distribution:")
print(df["condition"].value_counts())

# Download CSV to your device
files.download(csv_filename)


# 2. PREPARE DATA FOR AI
features = [
    "accel_x",
    "accel_y",
    "accel_z",
    "accel_magnitude",
    "vibration_do"
]

X = df[features].apply(pd.to_numeric, errors="coerce")
y = df["condition"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)


# 3. BUILD AND TRAIN RANDOM FOREST MODEL
model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("random_forest", RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        class_weight="balanced"
    ))
])

model.fit(X_train, y_train)

print("\nRANDOM FOREST MODEL TRAINED SUCCESSFULLY")


# 4. PREDICT AND EVALUATE
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred) * 100

print(f"\nModel accuracy on demo test split: {accuracy:.2f}%")
print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))


# 5. GRAPH 1: CONFUSION MATRIX
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    cmap="Blues"
)

plt.title("Random Forest - Confusion Matrix")
plt.tight_layout()
plt.show()


# 6. GRAPH 2: SENSOR FEATURE IMPORTANCE
rf = model.named_steps["random_forest"]

importance = pd.Series(
    rf.feature_importances_,
    index=features
).sort_values()

plt.figure(figsize=(8, 5))
importance.plot(kind="barh")
plt.xlabel("Feature Importance")
plt.ylabel("Sensor Feature")
plt.title("Random Forest - Sensor Feature Importance")
plt.tight_layout()
plt.show()


# 7. GRAPH 3: ACCELERATION MAGNITUDE OVER TIME
# Time is sorted for this graph because CSV rows were shuffled.
plot_df = df.sort_values("time_ms")

plt.figure(figsize=(10, 5))

for condition in ["Healthy", "Abnormal"]:
    subset = plot_df[plot_df["condition"] == condition]
    plt.plot(
        subset["time_ms"],
        subset["accel_magnitude"],
        label=condition,
        alpha=0.8
    )

plt.xlabel("Time (ms)")
plt.ylabel("Acceleration Magnitude (m/s²)")
plt.title("Acceleration Magnitude vs Time")
plt.legend()
plt.tight_layout()
plt.show()
```
